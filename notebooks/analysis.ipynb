```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from pathlib import Path
from src.hurst import hurst_rs, fractal_dimension_from_hurst
from src.periodogram import schuster_periodogram, top_periods

df = pd.read_csv("../data/sample.csv")   # має бути колонка "value"
x = df["value"].to_numpy()

H, det = hurst_rs(x, min_win=16, n_scales=8)
print("H:", H, " D:", 2-H)

f, P = schuster_periodogram(x, fs=1.0)  # якщо крок=1
plt.plot(f[1:], P[1:]); plt.xlabel("Hz"); plt.ylabel("Power"); plt.show()

periods, freqs, powers, _ = top_periods(x, fs=1.0, top_k=3)
print("Top periods:", periods)
```
